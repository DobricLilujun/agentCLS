```python
import matplotlib.pyplot as plt
import pandas as pd

data = {
    "Models": ["Llama-3.2-1B-Instruct", "Llama-3.2-3B-Instruct", "ModernBERT-base"],
    "EUR 50": [0.992, 0.999, 0.166],
    "EUR 150": [0.994, 0.998, 0.167],
    "EUR 1500":  [0.999, 0.998, 0.171],
    "LDD 50": [0.409, 0.615, 0.437],
    "LDD 150": [0.794, 0.828, 0.636],
    "LDD 1500": [0.890, 0.903, 0.811],
    "IE 50": [0.476, 0.462, 0.392],
    "IE 150": [0.643, 0.620, 0.333],
    "IE 1500": [0.863, 0.960, 0.408],
}
df = pd.DataFrame(data)

datasets = {"EUR": ["EUR 50", "EUR 150", "EUR 1500"],
            "LDD": ["LDD 50", "LDD 150", "LDD 1500"],
            "IE": ["IE 50", "IE 150", "IE 1500"]}

fig, axes = plt.subplots(1, 3, figsize=(15, 5), sharey=True)
legend_ax = axes[1] 
for ax, (dataset, columns) in zip(axes, datasets.items()):
    for _, row in df.iterrows():
        ax.plot([50, 150, 1500], row[columns], marker='o', label=row["Models"])
    ax.set_title(dataset)
    ax.set_xlabel("Training Samples")
    ax.set_ylabel("F1 Score")

legend_ax.legend()

plt.tight_layout()
plt.savefig('data_efficiency.svg', format='svg')
plt.show()
```
